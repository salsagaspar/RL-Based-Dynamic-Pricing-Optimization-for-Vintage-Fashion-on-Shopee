Analisis komprehensif 6 lembar kerja rantai pasok dan pesanan Shopee:
1. **Produk_Master**: Katalog SKU, biaya modal (HPP), harga acuan katalog, margin kotor.
2. **Pemasok**: Karakteristik vendor vintage (thrift/deadstock), lead time, tingkat cacat.
3. **Pembelian**: Purchase order & lead time aktual.
4. **Pesanan**: Transaksi penjualan, diskon, fee platform Shopee, dan margin neto.
5. **Retur**: Alasan pengembalian barang dan tingkat cacat.
6. **Persediaan**: Status stok awal vs akhir, safety stock, dan reorder point.
